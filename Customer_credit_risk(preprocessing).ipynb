```python
#load sql data:
import sqlite3
import pandas as pd

connection = sqlite3.connect("credit_risk.db")

with open("repayment_history.sql", "r") as file:
    sql_script = file.read()

# Remove MySQL-specific commands
sql_script = sql_script.replace(
    "CREATE DATABASE IF NOT EXISTS credit_risk_db;", ""
)

sql_script = sql_script.replace(
    "USE credit_risk_db;", ""
)

connection.executescript(sql_script)
connection.commit()

query = "SELECT * FROM repayment_history"

repayment_data = pd.read_sql(query, connection)

repayment_data.head()
```

```python
#missing indicator + random sample imputation:
original_income=loan_data.set_index("customer_id")["annual_income"]
combined_data['annual_income']=combined_data['customer_id'].map(original_income)
```

```python
combined_data["annual_income"].isnull().sum()
```

```python
combined_data["annual_income_missing"].value_counts()
```

```python
income_values=combined_data.loc[combined_data["annual_income"].notnull(),"annual_income"]
```

```python
import numpy as np
np.random.seed(42)
missing_count=combined_data["annual_income"].isnull().sum()
random_values=np.random.choice(income_values,size=missing_count,replace=True)
```

```python
combined_data.loc[combined_data["annual_income"].isnull(),"annual_income"]=random_values
```

```python
combined_data["annual_income"].isnull().sum()
```

```python
#knn imputer:
#restoring the original missing data
combined_data["age"] = combined_data["customer_id"].map(
    loan_data.set_index("customer_id")["age"]
)

combined_data["credit_score"] = combined_data["customer_id"].map(
    loan_data.set_index("customer_id")["credit_score"]
)
```

```python
#part D handle outliers :
combined_data.describe()
```

```python
#zscore method:
from scipy.stats import zscore
combined_data["Income_zscore"] = zscore(combined_data["annual_income"])
income_outliers=combined_data[combined_data["Income_zscore"].abs()>3]
```

```python
income_outliers[[
    "customer_id",
    "annual_income",
    "Income_zscore"
]]
len(income_outliers)
```

```python
combined_data = combined_data[combined_data["Income_zscore"].abs() <= 3].copy()
```

```python
combined_data.drop(columns=["Income_zscore"],inplace=True)
```

```python
#IQR method:
Q1=combined_data['loan_amount'].quantile(0.25)
Q3=combined_data['loan_amount'].quantile(0.75)
IQR=Q3-Q1
lower_limit= Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
combined_data=combined_data[(combined_data['loan_amount']>= lower_limit)&(combined_data['loan_amount']<= upper_limit)].copy()
```

```python
#percentile method:
lower_limit=combined_data['credit_score'].quantile(0.01)
upper_limit=combined_data['credit_score'].quantile(0.99)
combined_data=combined_data[(combined_data['credit_score']>=lower_limit)&(combined_data['credit_score']<=upper_limit)].copy()
```

```python
combined_data.shape
```

```python
#winsorization:
from scipy.stats.mstats import winsorize
combined_data['transaction_count']=winsorize(combined_data['transaction_count'],limits=[0.01,0.01])
```

```python
combined_data.shape
```

```python
#part E: feature Engineering
#handle variable type:

numeric_cols = combined_data.select_dtypes(include=['int64','float64']).columns
categoric_cols= combined_data.select_dtypes(include=['object']).columns
print("numerical variables:",numeric_cols)
print("categorical variables:",categoric_cols)
```

```python
#date time variable
combined_data['join_date']=pd.to_datetime(combined_data['join_date'])

combined_data['join_year']=combined_data['join_date'].dt.year
combined_data['join_month']=combined_data['join_date'].dt.month
combined_data['join_day']=combined_data['join_date'].dt.day
combined_data['join_weekday']=combined_data['join_date'].dt.day_name()
```

```python
combined_data.head()
```

```python
#encoding:
#ordinal encoding:
combined_data['education_level'].unique()
```

```python
education_map= {'Primary':1,'Secondary':2,'Post-Graduate':3,'Graduate':4}
combined_data['education_level']=combined_data['education_level'].map(education_map)
```

```python
#label encoding:
from sklearn.preprocessing import LabelEncoder
label_encoder= LabelEncoder()
combined_data['gender']= label_encoder.fit_transform(combined_data['gender'])
```

```python
combined_data['gender'].value_counts()
```

```python
#one hot encoding :
combined_data= pd.get_dummies(combined_data,columns=['region','loan_purpose'],dtype=int)
```

```python
combined_data.head()
```

```python
#encoding numerical features:
#binning:
bins=[0,30000,60000,100000,float('inf')]
labels=['Low','Medium','High','Very High']
combined_data['Income_group']=pd.cut(combined_data['annual_income'],bins=bins,labels=labels)
```

```python
#Binarization:
combined_data['high_credit_score']=(combined_data['credit_score']>700).astype(int)
```

```python
combined_data[['credit_score','high_credit_score']].head()
```

```python
#Quantile binning:
combined_data['Income_quantile']=pd.qcut(combined_data['annual_income'],q=4,labels=['Q1','Q2','Q3','Q4'])
```

```python
combined_data[['annual_income','Income_quantile']].head()
```

```python
#Kmean binning:
from sklearn.cluster import KMeans
kmeans=KMeans(n_clusters=4, random_state=42,n_init=10)
combined_data['income_kmeans']=kmeans.fit_predict(combined_data[['annual_income']])
```

```python
combined_data[['annual_income','income_kmeans']].head()
```

```python
#part F : Feature scaling
#standardization:
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
combined_data['annual_income']=scaler.fit_transform(combined_data[['annual_income']])
```

```python
combined_data['annual_income'].head()
```

```python
#Normalization:
from sklearn.preprocessing import Normalizer
normalizer = Normalizer()
combined_data['loan_amount']=normalizer.fit_transform(combined_data[['loan_amount']])
```

```python
combined_data['loan_amount'].head()
```

```python
#min max scaling
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
combined_data['credit_score']=scaler.fit_transform(combined_data[['credit_score']])
```

```python
combined_data['credit_score'].head()
```

```python
#maxabs scaling
from sklearn.preprocessing import MaxAbsScaler
scaler=MaxAbsScaler()
combined_data['transaction_count']=scaler.fit_transform(combined_data[['transaction_count']])
```

```python
combined_data['transaction_count'].head()
```

```python
#robust scaling:
from sklearn.preprocessing import RobustScaler
scaler=RobustScaler()
combined_data['repayment_history']=scaler.fit_transform(combined_data[['repayment_history']])
```

```python
combined_data['repayment_history'].head()
```

```python
#part G feature construction and transformation:
#function Transformer: log transform
from sklearn.preprocessing import FunctionTransformer
log_transformer=FunctionTransformer(np.log1p)
combined_data['income_log']=log_transformer.fit_transform(combined_data[['annual_income']])
```

```python
#reciprocal transformer
reciprocal_transformer= FunctionTransformer(lambda x:1/x)
combined_data['loan_amount_reciprocal'] = reciprocal_transformer.fit_transform(combined_data[['loan_amount']])
```

```python
#square root :
sqrt_transformer = FunctionTransformer(np.sqrt)
combined_data['transaction_count_sqrt'] = sqrt_transformer.fit_transform(combined_data[['transaction_count']])
```

```python
#powertransformer : box cox
from sklearn.preprocessing import PowerTransformer
boxcox_transformer=PowerTransformer(method='box-cox')
combined_data['age']= boxcox_transformer.fit_transform(combined_data[['age']] )
```

```python
#powertransformer : yeo -johnson
from sklearn.preprocessing import PowerTransformer
yeojohnson_transformer=PowerTransformer(method='yeo-johnson')
combined_data['income_johnson']= yeojohnson_transformer.fit_transform(combined_data[['annual_income']] )
```

```python
combined_data.head()
```

```python
#column transformer:
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
preprocessor= ColumnTransformer(transformers=[('num',StandardScaler(),['economic_index']),
                                             ('cat',OneHotEncoder(handle_unknown='ignore'),['employment_type'])])

transformed_data= preprocessor.fit_transform(combined_data)
```
